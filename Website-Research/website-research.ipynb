```python
# import modules 
from openai import OpenAI
import requests
from bs4 import BeautifulSoup
```

```python
# webscraping 

headers = {
    "User-Agent": "LLMEngineeringProject/1.0 (chandrabadrinathreddy@gmail.com)"
}

url = "https://en.wikipedia.org/wiki/OpenAI"

response = requests.get(url, headers=headers)

html = response.text
print(html)

soup = BeautifulSoup(html, 'html.parser')
title = soup.title.string

text = soup.get_text()

for tag in soup(["script", "style"]):
    tag.decompose()

text = soup.get_text(separator="\n", strip=True)
```

```python
# promts
system_promt = "your are my research assistant"
user_promt = f"""I want complete research about this website. Please find the below response format
1. Overview
2. Key Facts
3. Important People
4. Timeline
5. Summary

{text}
 """
messages = [{"role": "system", "content": system_promt}, {"role": "user", "content": user_promt}]
```

```python
# making llm call to the phi4

MODEL = "phi4-mini:latest"
BASE_URL = "http://localhost:11434/v1/"

phi4 = OpenAI(base_url=BASE_URL, api_key="everything")

response = phi4.chat.completions.create(
    model = MODEL,
    messages = messages
)

response.choices[0].message.content
```
